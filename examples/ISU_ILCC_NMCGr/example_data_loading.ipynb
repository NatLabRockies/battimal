```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from ISUILCC_NMCGr_Data import Dataset_ISUILCC_NMCGr
```

```python
data = Dataset_ISUILCC_NMCGr(is_short=True)
```

```python
data.cell_extracts_and_metadata_to_df()
```

```python
plt.plot(data[0][0].Time_s, data[0][0].Voltage_V)
```

```python
cell_data = data[0]
fig, ax = plt.subplots(2, 6, figsize=(16, 6))
ax = ax.ravel()
for QV_dQdV_dVdQ, measurement_type in zip(cell_data.measurement_extracts['charge_QV_dQdV_dVdQ'], cell_data.measurement_type):
    if measurement_type == 'RPT C_5':
        ax[0].plot(QV_dQdV_dVdQ['Voltage_V'], QV_dQdV_dVdQ['Capacity_Ah'])
        ax[1].plot(QV_dQdV_dVdQ['Voltage_V'], QV_dQdV_dVdQ['dQdV_Ah/V'])
        ax[2].plot(QV_dQdV_dVdQ['Capacity_Ah'], QV_dQdV_dVdQ['dVdQ_V/Ah'])
for QV_dQdV_dVdQ, measurement_type in zip(cell_data.measurement_extracts['discharge_QV_dQdV_dVdQ'], cell_data.measurement_type):
    if measurement_type == 'RPT C_5':
        ax[3].plot(QV_dQdV_dVdQ['Voltage_V'], QV_dQdV_dVdQ['Capacity_Ah'])
        ax[4].plot(QV_dQdV_dVdQ['Voltage_V'], QV_dQdV_dVdQ['dQdV_Ah/V'])
        ax[5].plot(QV_dQdV_dVdQ['Capacity_Ah'], QV_dQdV_dVdQ['dVdQ_V/Ah'])
for QV_dQdV_dVdQ, measurement_type in zip(cell_data.measurement_extracts['charge_QV_dQdV_dVdQ'], cell_data.measurement_type):
    if measurement_type == 'RPT C_2':
        ax[6].plot(QV_dQdV_dVdQ['Voltage_V'], QV_dQdV_dVdQ['Capacity_Ah'])
        ax[7].plot(QV_dQdV_dVdQ['Voltage_V'], QV_dQdV_dVdQ['dQdV_Ah/V'])
        ax[8].plot(QV_dQdV_dVdQ['Capacity_Ah'], QV_dQdV_dVdQ['dVdQ_V/Ah'])
for QV_dQdV_dVdQ, measurement_type in zip(cell_data.measurement_extracts['discharge_QV_dQdV_dVdQ'], cell_data.measurement_type):
    if measurement_type == 'RPT C_2':
        ax[9].plot(QV_dQdV_dVdQ['Voltage_V'], QV_dQdV_dVdQ['Capacity_Ah'])
        ax[10].plot(QV_dQdV_dVdQ['Voltage_V'], QV_dQdV_dVdQ['dQdV_Ah/V'])
        ax[11].plot(QV_dQdV_dVdQ['Capacity_Ah'], QV_dQdV_dVdQ['dVdQ_V/Ah'])

ax[1].set_ylim(0, 1)
ax[4].set_ylim(-1, 0)
ax[7].set_ylim(0, 1)
ax[10].set_ylim(-1, 0)

ax[2].set_ylim(0, 10)
ax[5].set_ylim(-10, 0)
ax[8].set_ylim(0, 10)
ax[11].set_ylim(-10, 0)

plt.tight_layout()
```
