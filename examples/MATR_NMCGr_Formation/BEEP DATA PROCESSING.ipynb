```python
fig, ax = plt.subplots(figsize=(10, 6))
ax_2 = ax.twinx()
for cycle in structure.diagnostic_data['cycle_index'].unique():
    cycle_data = structure.diagnostic_data[structure.diagnostic_data['cycle_index'] == cycle]
    if 'rpt' in cycle_data['cycle_type'].iloc[0]:
        time = cycle_data['test_time'] - cycle_data['test_time'].iloc[0]
        ax.plot(time/3600, cycle_data['voltage'])
        ax_2.plot(time/3600, cycle_data['discharge_capacity'], color='orange')
ax.set_xlabel('Time (hr)')
ax.set_ylabel('Voltage (V)')
ax_2.set_ylabel('Discharge Capacity (Ah)')
```

```python
folder = os.path.join('raw', 'Aging')
files = os.listdir(folder)

for file in files:
    fname = file[:-5]
    print(fname)
    structure = auto_load_processed(os.path.join(folder, file))
    structure.structured_data.to_pickle(os.path.join('raw', 'Aging (extracted from BEEP format)', f'{fname}_agingCycles_data.pkl'))
    structure.structured_summary.to_pickle(os.path.join('raw', 'Aging (extracted from BEEP format)', f'{fname}_agingCycles_summary.pkl'))
    structure.diagnostic_data.to_pickle(os.path.join('raw', 'Aging (extracted from BEEP format)', f'{fname}_rpt_data.pkl'))
    structure.diagnostic_summary.to_pickle(os.path.join('raw', 'Aging (extracted from BEEP format)', f'{fname}_rpt_summary.pkl'))
    del structure
```
